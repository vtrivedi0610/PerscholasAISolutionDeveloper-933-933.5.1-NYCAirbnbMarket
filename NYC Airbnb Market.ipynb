## What share of listings are entire home/apt vs. private room vs. others?

## Do some room types dominate certain price bands?

• 
	• Budget & Economy Tiers ($0–$100/night): Private rooms heavily dominate the sub-$100 market (typically capturing 75%–85%+ of listings in these bands), joined by almost all shared rooms. Travelers shopping under $100 in NYC are almost exclusively booking private rooms.
	• Mid-Range Tier ($101–$175/night): Serves as the primary crossover zone where entire homes/apartments take over the majority share (~65%–75%), though upscale private rooms (often in Manhattan or prime Brooklyn) still compete here.
	• Upscale & Luxury Tiers ($176+/night): Entire homes/apartments completely dominate (capturing 90%+ of inventory above $175), as very few private rooms command full-apartment rates.
